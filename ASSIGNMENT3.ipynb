```python
import numpy as np
import matplotlib.pyplot as plt

# Generate sample grayscale image
height, width = 180, 240
y, x = np.mgrid[0:height, 0:width]

gray_image = np.clip(40 + 0.6*x + 0.4*y, 0, 255).astype(np.uint8)

yy, xx = np.ogrid[:height, :width]
gray_image[(xx-75)**2 + (yy-75)**2 <= 35**2] = 230
gray_image[35:85, 145:205] = 20

# Create binary image
binary_image = np.where(gray_image >= 128, 255, 0).astype(np.uint8)


# PART 1: IMAGE PADDING

border_width = 15
border_color = 255

# Grayscale image padding
gray_padded = np.pad(
    gray_image,
    ((border_width, border_width), (border_width, border_width)),
    mode='constant',
    constant_values=border_color
)

# Binary image padding
binary_padded = np.pad(
    binary_image,
    ((border_width, border_width), (border_width, border_width)),
    mode='constant',
    constant_values=border_color
)

# Display grayscale padding
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(gray_image, cmap='gray', vmin=0, vmax=255)
plt.title("Original Grayscale Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(gray_padded, cmap='gray', vmin=0, vmax=255)
plt.title("Grayscale Image with Border")
plt.axis('off')

plt.tight_layout()
plt.show()

# Display binary padding
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(binary_image, cmap='gray', vmin=0, vmax=255)
plt.title("Original Binary Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(binary_padded, cmap='gray', vmin=0, vmax=255)
plt.title("Binary Image with Border")
plt.axis('off')

plt.tight_layout()
plt.show()


# PART 2: IMAGE COMPLEMENT

# Grayscale complement
gray_complement = 255 - gray_image

# Binary complement
binary_complement = 255 - binary_image

# Display grayscale complement
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(gray_image, cmap='gray', vmin=0, vmax=255)
plt.title("Original Grayscale Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(gray_complement, cmap='gray', vmin=0, vmax=255)
plt.title("Grayscale Image Complement")
plt.axis('off')

plt.tight_layout()
plt.show()

# Display binary complement
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(binary_image, cmap='gray', vmin=0, vmax=255)
plt.title("Original Binary Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(binary_complement, cmap='gray', vmin=0, vmax=255)
plt.title("Binary Image Complement")
plt.axis('off')

plt.tight_layout()
plt.show()
```
