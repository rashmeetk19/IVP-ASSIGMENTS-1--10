```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Generate a sample grayscale image
height, width = 180, 240

y, x = np.mgrid[0:height, 0:width]

gray_image = np.clip(
    35 + 0.65*x + 0.35*y, 0, 255
).astype(np.uint8)

# Add shapes to the image
yy, xx = np.ogrid[:height, :width]

gray_image[(xx-75)**2 + (yy-75)**2 <= 38**2] = 230
gray_image[35:85, 145:205] = 25


# Display original grayscale image
plt.figure(figsize=(6, 4))
plt.imshow(gray_image, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis('off')
plt.show()


# CASE 1: Mean intensity as threshold

threshold = np.mean(gray_image)

binary_image1 = np.where(
    gray_image >= threshold, 255, 0
).astype(np.uint8)

print("Case 1: Mean Intensity Threshold")
print("Threshold value:", threshold)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(gray_image, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(binary_image1, cmap='gray')
plt.title("Binary Image - Mean Threshold")
plt.axis('off')

plt.tight_layout()
plt.show()


# CASE 2: Threshold value entered by user

user_threshold = int(input("Enter threshold value (0-255): "))

if not 0 <= user_threshold <= 255:
    raise ValueError("Threshold must be between 0 and 255")

binary_image2 = np.where(
    gray_image >= user_threshold, 255, 0
).astype(np.uint8)

print("Case 2: User Defined Threshold")
print("Threshold value:", user_threshold)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(gray_image, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(binary_image2, cmap='gray')
plt.title("Binary Image - User Threshold")
plt.axis('off')

plt.tight_layout()
plt.show()
```
